# Q1,2,3

```python
""" Data : Use the ‘wti-daily.csv’ and ‘brent-daily.csv’ files into a data frame
in which the Date column is treated as a datetime value and is set to be
the index.
Q1. Plot the average price trend of oil from 1992-2002 from wtidaily.csv?
Q2. Plot the average price trend of oil from 1992 -2002
Q3. Compare the both the average prices of a barrel of oil from 1992-
2002, indicate significant differences though markers."""

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive
import matplotlib.dates as mdates

drive.mount('/content/drive')

# Load CSV files
try:
    wti_df = pd.read_csv('/content/asgmt data/wti-daily.csv')
    brent_df = pd.read_csv('/content/asgmt data/brent-daily.csv')

    # Convert 'Date' column to datetime and set as index
    wti_df['Date'] = pd.to_datetime(wti_df['Date'])
    wti_df.set_index('Date', inplace=True)
    brent_df['Date'] = pd.to_datetime(brent_df['Date'])
    brent_df.set_index('Date', inplace=True)

    # Filter data for the specified date range (1992-2002)
    start_date = '1992-01-01'
    end_date = '2002-12-31'
    wti_filtered_df = wti_df[(wti_df.index >= start_date) & (wti_df.index <= end_date)].copy()
    brent_filtered_df = brent_df[(brent_df.index >= start_date) & (brent_df.index <= end_date)].copy()

    # Ensure the 'Price' column is numeric
    wti_filtered_df['Price'] = pd.to_numeric(wti_filtered_df['Price'], errors='coerce')
    brent_filtered_df['Price'] = pd.to_numeric(brent_filtered_df['Price'], errors='coerce')

    # Resample to monthly average
    wti_monthly_avg = wti_filtered_df['Price'].resample('ME').mean()
    brent_monthly_avg = brent_filtered_df['Price'].resample('ME').mean()

    # Create a figure with one row and three subplots
    fig, axes = plt.subplots(1, 3, figsize=(20, 5))

    # Q1: Plot WTI daily price trend in the first subplot
    axes[0].plot(wti_filtered_df.index, wti_filtered_df['Price'])
    axes[0].set_title('WTI Daily Price (1992-2002)')
    axes[0].set_xlabel('Date')
    axes[0].set_ylabel('Price')
    axes[0].grid(True)
    axes[0].xaxis.set_major_locator(mdates.YearLocator(1))
    axes[0].xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

    # Q2: Plot Brent daily price trend in the second subplot
    axes[1].plot(brent_filtered_df.index, brent_filtered_df['Price'], color='orange')
    axes[1].set_title('Brent Daily Price (1992-2002)')
    axes[1].set_xlabel('Date')
    axes[1].set_ylabel('Price')
    axes[1].grid(True)
    axes[1].xaxis.set_major_locator(mdates.YearLocator(1))
    axes[1].xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

    # Q3: Compare the average prices (Monthly) in the third subplot
    axes[2].plot(wti_monthly_avg.index, wti_monthly_avg.values, label='WTI (Monthly Avg)', marker='o', markersize=4, linestyle='-')
    axes[2].plot(brent_monthly_avg.index, brent_monthly_avg.values, label='Brent (Monthly Avg)', marker='x', markersize=4, linestyle='-')
    axes[2].set_title('Comparison of WTI and Brent Monthly Average Prices (1992-2002)')
    axes[2].set_xlabel('Date')
    axes[2].set_ylabel('Price')
    axes[2].legend()
    axes[2].grid(True)
    axes[2].xaxis.set_major_locator(mdates.YearLocator(1))
    axes[2].xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

    # Add annotation to the third subplot
    axes[2].annotate('Significant Difference 1',
                     xy=(pd.to_datetime('1993-06-01'), 21),
                     xytext=(pd.to_datetime('1993-07-01'), 27),
                     arrowprops=dict(facecolor='black', shrink=0.05),
                     horizontalalignment='center', verticalalignment='bottom')
    axes[2].annotate('Significant Difference 2',
                     xy=(pd.to_datetime('2001-01-01'), 24),
                     xytext=(pd.to_datetime('2001-06-01'), 15),
                     arrowprops=dict(facecolor='black', shrink=0.05),
                     horizontalalignment='center', verticalalignment='bottom')

    plt.tight_layout()
    plt.show()


except FileNotFoundError as e:
    print(f"Error loading file: {e}. Please check the file path.")
except KeyError as e:
    print(f"Error processing data: {e}. Please ensure required columns exist.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

# Q4,5

```python
# Q4 & Q5: City-wise distribution of sales and payment method impact
# Data: sales.xlsx (place in same folder as this notebook)
import pandas as pd
import matplotlib.pyplot as plt

drive.mount('/content/drive')

try:
    sales = pd.read_excel('/content/sales.xlsx')

    # Explicit column names (as provided in the header)
    city_col = 'City'
    sales_col = 'Total'
    pay_col = 'Payment'

    # Ensure sales column is numeric
    sales[sales_col] = pd.to_numeric(sales[sales_col], errors='coerce')

    # Q4: City-wise total sales
    city_sum = sales.groupby(city_col)[sales_col].sum().sort_values(ascending=False)
    top_city = city_sum.index[0] if len(city_sum) > 0 else None

    # Q5: Payment method contribution to sales
    pay_sum = sales.groupby(pay_col)[sales_col].sum().sort_values(ascending=False)
    top_payment = pay_sum.index[0] if len(pay_sum) > 0 else None

    # Print short answers
    if top_city is not None:
        print(f'Q4 -> Top city by total sales: {top_city} ({city_sum.iloc[0]:.2f})')
    else:
        print('Q4 -> No city sales data available')

    if top_payment is not None:
        print(f'Q5 -> Top payment method by total sales: {top_payment} ({pay_sum.iloc[0]:.2f})')
    else:
        print('Q5 -> Payment method column not found or no data')

    # Plotting
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # City plot
    bars = axes[0].bar(city_sum.index.astype(str), city_sum.values, color='skyblue')
    axes[0].set_title('Total Sales Per City')
    axes[0].set_xlabel('City')
    axes[0].set_ylabel('Total Sales')
    axes[0].tick_params(axis='x', rotation=45)

    # annotate max city by finding the bar index for top_city (no loop)
    if top_city is not None and len(bars) > 0:
        labels = list(city_sum.index.astype(str))
        try:
            pos = labels.index(str(top_city))
            b = bars[pos]
            x = b.get_x() + b.get_width() / 2
            max_val = city_sum.iloc[pos]
            axes[0].annotate(f'Max:({max_val:.2f})', xy=(x, max_val), xytext=(x, max_val * 1.02), ha='center')
        except ValueError:
            pass

    # Payment method plot
    bars2 = axes[1].bar(pay_sum.index.astype(str), pay_sum.values, color='gray')
    axes[1].set_title('Total Sales Per Payment Method')
    axes[1].set_xlabel('Payment Method')
    axes[1].set_ylabel('Total Sales')
    axes[1].tick_params(axis='x', rotation=45)

    # annotate max payment method by finding the bar index for top_payment (no loop)
    if top_payment is not None and len(bars2) > 0:
        pay_labels = list(pay_sum.index.astype(str))
        try:
            pos2 = pay_labels.index(str(top_payment))
            b2 = bars2[pos2]
            x2 = b2.get_x() + b2.get_width() / 2
            max_val2 = pay_sum.iloc[pos2]
            axes[1].annotate(f'Max:({max_val2:.2f})', xy=(x2, max_val2), xytext=(x2, max_val2 * 1.02), ha='center')
        except ValueError:
            pass

    plt.tight_layout()
    plt.show()

except FileNotFoundError:
    print("sales.xlsx not found. Please place it in the notebook directory.")
except Exception as e:
    print('Error processing sales.xlsx:', e)
```

```python

import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np

# Define the path to the image file
image_path = '/content/asgmt data/2.jpg'


try:
    # Read the image into a NumPy array
    img_array = mpimg.imread(image_path)
    print(f"Image loaded successfully. Shape: {img_array.shape}, Data type: {img_array.dtype}")

    # Visualize the image
    plt.imshow(img_array)
    plt.title('Visualizing 2.jpg')
    plt.axis('off') # Hide axes ticks and labels
    plt.show()

except FileNotFoundError:
    print(f"Error: The file '{image_path}' was not found. Please check the path.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
from skimage.color import rgb2gray

gray_img = rgb2gray(img_array)
```

```python
colormaps = ['gray', 'viridis', 'hot', 'magma', 'coolwarm', 'hsv']

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten() # Flatten the 2x3 array of axes for easier iteration

for i, cmap_name in enumerate(colormaps):
    ax = axes[i]
    ax.imshow(gray_img, cmap=cmap_name)
    ax.set_title(f'Colormap: {cmap_name}')
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from PIL import Image, ImageFilter, ImageEnhance

# Create a figure
fig = plt.figure(figsize=(15, 12))

# Define the GridSpec for a 3x3 grid
gs = gridspec.GridSpec(3, 3, figure=fig)

# First row: 3 subplots
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[0, 2])

# Second row: 1 subplot spanning all 3 columns
ax4 = fig.add_subplot(gs[1, :])

# Third row: 2 subplots, centered by using the first and third columns
ax5 = fig.add_subplot(gs[2, 0])
ax6 = fig.add_subplot(gs[2, 2])

# Collect all axes for easier iteration
axes = [ax1, ax2, ax3, ax4, ax5, ax6]

# Convert img_array to a PIL Image for processing
pil_img = Image.fromarray(img_array)

# Define image processing operations and their titles
operations = [
    (pil_img.filter(ImageFilter.SHARPEN), 'Sharpened Image'),
    (pil_img.filter(ImageFilter.GaussianBlur(radius=4)), 'Gaussian Blur'),
    (ImageEnhance.Contrast(pil_img).enhance(2), 'Contrast Enhanced'),
    (ImageEnhance.Brightness(pil_img).enhance(2), 'Brightness Enhanced'),
    (pil_img.filter(ImageFilter.FIND_EDGES), 'Edge Detection'),
    (pil_img.convert('L'), 'Grayscale')
]

# Display processed images in each subplot
for i, (processed_img, title) in enumerate(operations):
    ax = axes[i]
    ax.imshow(processed_img, cmap='gray' if title == 'Grayscale' else None)
    ax.set_title(title)
    ax.axis('off') # Turn off axes for cleaner presentation

plt.tight_layout()
plt.show()
```
